```python
kb = KnowledgeBase()
kb.assertz("parent(john, mary).")
kb.assertz("parent(mary, alice).")
kb.assertz("ancestor(X, Y) :- parent(X, Y).")
kb.assertz("ancestor(X, Y) :- parent(X, Z), ancestor(Z, Y).")

print("=== Verbose + Abduction Query ===")
answers = kb.query("ancestor(john, Who).", abduction=True)
print("\nAnswers:")
for a in answers:
    print(a)

print("\n=== Retract parent(mary, alice) ===")
kb.retract("parent(mary, alice).")
answers2 = kb.query("ancestor(john, Who).", abduction=True)
print("\nAnswers after retraction:")
for a in answers2:
    print(a)
```
